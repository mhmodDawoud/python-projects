## input user :)

```python

weather = input("Enter weather: ").lower().strip()
time_of_day = input("Enter time of day: ").lower().strip()
```

## print the input

```python
print(weather)
print("-" * 20)
print(time_of_day)
```

```python
def decide_outfit(weather, time_of_day):
    if weather == "rainy"  and  time_of_day == "morning" or "afternoon":
        print("Wear a waterproof jacket.")
    elif  weather == "sunny"  and time_of_day ==  "morning" or "afternoon":
        print("Wear light clothing and a sunhat.")
    elif weather == "rainy"  and  time_of_day == "night":
        print("Wear warm waterproof clothing and carry an umbrella.")
    elif weather == "snowy" and time_of_day == "morning" or "afternoon" or "evening":
        print("Wear heavy winter clothing.")
    else:
        print(" Check your input :) ")    
```

```python
decide_outfit(weather , time_of_day)
```
